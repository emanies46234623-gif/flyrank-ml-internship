## 2. The question: decision, action, cost of a wrong call

Unit of analysis (grain): One row = one content page (`content_id`), from the starter 
dataset's anonymized 30,000-page snapshot.

The decision: Which pages a content reviewer, with capacity to review roughly 50 pages per 
cycle, should look at first.

The action: A reviewer manually inspects the top-ranked pages and decides whether to refresh, 
rewrite metadata, expand thin content, or leave the page alone.

The cost of a wrong call:
- False positive (flagged, but not actually a problem): wasted reviewer time — roughly 
  30-60 minutes spent reviewing a page that didn't need attention.
- False negative (missed, but actually declining): a real traffic/visibility problem goes 
  unaddressed until the next review cycle, potentially weeks later.

Why data/ML helps at all: With thousands of pages and capacity for only ~50 reviews per 
cycle, picking pages by a transparent, evidence-based ranking beats picking randomly or by gut 
feel — the starter pipeline already shows a learned model roughly triples the hit-rate of a 
simple hand-written rule at the top of the queue (Precision@50: 0.240 → 0.740).


## 3. Quick look at the data (2-3 real numbers)

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: how many pages actually qualify for review
qualifying = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
print(f"Total pages in starter dataset: {df.shape[0]}")
print(f"Pages qualifying for review (impressions_90d > 0, content_age_days >= 90): {qualifying.shape[0]}")

# Number 2: how many of those are declining under the starter's proxy label
declining_rate = (qualifying["trend_direction"] == "down").mean()
print(f"Declining rate among qualifying pages: {declining_rate:.3f}")

# Number 3: evidence the ranking problem is worth solving (from your own pipeline run)
import json
res = json.load(open("outputs/model_results.json"))
base = res["baseline"]["baseline_precision_at_50"]
rf = res["models"]["random_forest"]["precision_at_50"]
print(f"Baseline rule Precision@50: {base:.3f}")
print(f"Random forest Precision@50: {rf:.3f}  (~{rf/base:.1f}x the baseline)")
```

Of the pages qualifying for review under a simple volume/age filter, a meaningful share are 
currently labeled declining under the starter's proxy definition. More importantly, the pipeline
