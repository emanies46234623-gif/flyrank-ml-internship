## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
## 1. My Lane as an ML Task

**Task type: Classification**

My ML task is to classify each content item according to its observed performance trend, with particular attention to identifying content that is trending down.

The target variable is `trend_direction`. The practical goal is to identify content at risk of declining performance so the content team can prioritize those items for review and potential refresh.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
## 2. Target or Proxy

**Target: `trend_direction`**

The target is an observed outcome already present in the starter dataset. It represents the direction of recent content performance and includes categories such as `down`, `stable`, `up`, `new`, and `flat`.

For the content-refresh use case, the most important outcome is identifying content classified as `down`, because these items may need review or refresh.

I will treat `trend_direction` as the observed target rather than creating a manually defined refresh-priority label.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
## 3. Success Metric

**Success metric: F1 score for the `down` class**

I will use F1 score for the `down` class because the main business use is identifying content that is trending down for review and potential refresh. F1 balances precision and recall, helping avoid both missing too many declining items and incorrectly flagging too many items.

As an initial success target, I will aim for an F1 score of **0.70 or higher** on a held-out evaluation set. This is a working target that should be compared with a simple baseline before deciding whether the model is useful.

A stronger score is valuable only if it leads to better identification of content that the team can actually review and refresh.
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
## 5. Why ML Beats a Fixed Rule

A simple fixed rule could identify declining content using one manually chosen condition, such as marking every item with a negative recent trend as a refresh candidate.

However, content performance can be influenced by multiple factors, including traffic history, search visibility, content age, freshness, engagement, and content characteristics. A single threshold may not capture the different combinations of these signals associated with different trend outcomes.

ML is useful because it can learn relationships across multiple features and identify patterns associated with different trend outcomes. This can provide more flexible decision support than relying only on one or two hand-written thresholds.

The output supports a real content action: helping the content team identify content that may need review and potential refresh.
```
