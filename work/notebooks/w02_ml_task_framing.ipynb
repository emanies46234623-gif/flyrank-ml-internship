## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
## 1. My Lane as an ML Task

**Task type: Ranking**

My ML task is to rank content items by their priority for review and potential refresh.

The goal is to help a content team decide which existing content should be reviewed first by combining signals such as recent traffic changes, search performance, engagement, content age, freshness, and other available content-level features.

The output will be an ordered list of content items, from highest refresh priority to lowest.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
## 2. Target or Proxy

**Target / Proxy: Content refresh priority score**

The starter dataset does not contain a direct ground-truth label for whether a piece of content should be refreshed. Therefore, I would use a proxy priority score constructed from observable signals that indicate refresh opportunity.

The proxy would combine signals such as `trend_pct`, `impressions_last_30d`, `clicks_last_30d`, `avg_position`, engagement, `content_age_days`, and `days_since_last_update`.

The proxy represents how strongly a content item appears to deserve review or refresh attention.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
## 3. Success Metric

**Success metric: NDCG@K**

I would use Normalized Discounted Cumulative Gain at K (NDCG@K) because the goal is to place the highest-priority content items near the top of the ranked list.

A successful model should consistently place the content items with the strongest refresh opportunity in the top K positions.

The value of the ranking should ultimately be reflected in a better prioritization of the content team's review and refresh workload.
```

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
## 4. The Unit of Analysis

**One row = one content item.**

Each row represents one individual content item and contains its search, traffic, engagement, freshness, age, and trend-related attributes.

The model would produce one refresh-priority score for each content item.
```
