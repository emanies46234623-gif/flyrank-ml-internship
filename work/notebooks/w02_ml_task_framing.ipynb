## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
## 1. My Lane as an ML Task

**Task type: Classification**

My ML task is to classify each content item according to its observed performance trend, with particular attention to identifying content that is trending down.

The target variable is `trend_direction`. The practical goal is to identify content at risk of declining performance so the content team can prioritize those items for review and potential refresh.
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
## 2. Target or Proxy

**Target: `trend_direction`**

The target is an observed outcome already present in the starter dataset. It represents the direction of recent content performance and includes categories such as `down`, `stable`, `up`, `new`, and `flat`.

For the content-refresh use case, the most important outcome is identifying content classified as `down`, because these items may need review or refresh.

I will treat `trend_direction` as the observed target rather than creating a manually defined refresh-priority label.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
## 3. Success Metric

**Success metric: F1 score**

I will use F1 score because the goal is to identify content items that are trending down while balancing precision and recall.

A high F1 score means the model is doing a good job of identifying the relevant declining-content cases without producing excessive false positives.

For the content team, better identification of declining content would improve the prioritization of items for review and potential refresh.
```

## 4. The unit of analysis, as a real dataframe

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
## 5. Why ML Beats a Fixed Rule

A simple fixed rule could identify declining content using one manually chosen condition, such as marking every item with a negative recent trend as a refresh candidate.

However, content performance is influenced by multiple factors, including previous traffic, search visibility, content age, freshness, engagement, and content characteristics. A single threshold would not capture the different combinations of these signals that may be associated with future performance trends.

ML is useful because it can learn relationships between multiple features and identify patterns associated with different trend outcomes. This allows the content team to use a model-based prediction to support prioritization rather than relying only on one manually selected threshold.

The output supports a real content action: identifying content that may require review and potential refresh.
```
